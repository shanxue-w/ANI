```python
import sympy as sp
from IPython.display import display
from sympy import Eq

# 定义符号变量
t = sp.Symbol('t')
# 直接代入数值，不再作为符号变量
# m1_val, m2_val = 0.4795663058757782, 0.3798398971557617
# L1_val, L2_val = 0.13726119697093964, 0.11183536052703857
m1_val, m2_val = sp.symbols('m1 m2')
L1_val, L2_val = sp.symbols('L1 L2')
g_val = 10

# 定义角度和角速度作为时间 t 的函数
theta1 = sp.Function('theta1')(t)
theta2 = sp.Function('theta2')(t)

# 定义角速度和角加速度
theta1_dot = sp.diff(theta1, t)
theta2_dot = sp.diff(theta2, t)
theta1_ddot = sp.diff(theta1_dot, t)
theta2_ddot = sp.diff(theta2_dot, t)

# 计算动能 T
T1 = sp.Rational(1, 2) * m1_val * (L1_val * theta1_dot)**2
T2 = sp.Rational(1, 2) * m2_val * (
    (L1_val * theta1_dot)**2 + (L2_val * theta2_dot)**2 + 2 * L1_val * L2_val * theta1_dot * theta2_dot * sp.cos(theta1 - theta2)
)
T = T1 + T2

# 计算势能 V
V = -((m1_val + m2_val) * g_val * L1_val * sp.cos(theta1) + m2_val * g_val * L2_val * sp.cos(theta2))

# 计算拉格朗日量 L
L = T - V

# 使用欧拉-拉格朗日方程
eq1_lhs = sp.diff(sp.diff(L, theta1_dot), t) - sp.diff(L, theta1)
eq2_lhs = sp.diff(sp.diff(L, theta2_dot), t) - sp.diff(L, theta2)
display(eq1_lhs)

# 提取 theta1_ddot 和 theta2_ddot 的系数
A11 = sp.simplify(sp.diff(eq1_lhs, theta1_ddot))
A12 = sp.simplify(sp.diff(eq1_lhs, theta2_ddot))
A21 = sp.simplify(sp.diff(eq2_lhs, theta1_ddot))
A22 = sp.simplify(sp.diff(eq2_lhs, theta2_ddot))

# 提取不含 theta1_ddot 和 theta2_ddot 的项
B1 = sp.simplify(eq1_lhs - A11*theta1_ddot - A12*theta2_ddot)
B2 = sp.simplify(eq2_lhs - A21*theta1_ddot - A22*theta2_ddot)

# 构建矩阵
M = sp.Matrix([[A11, A12], [A21, A22]])
RHS = sp.Matrix([B1, B2])

display(M, RHS)

# 求解线性方程组 M * X = -RHS
solutions = M.solve(-RHS)

theta1_ddot_expr = solutions[0]
theta2_ddot_expr = solutions[1]

print("--- 无阻尼双摆一阶 ODE 系统 (dx/dt) ---")
display(Eq(sp.Symbol(r'd( \theta_{1} )/dt'), theta1_dot))
display(Eq(sp.Symbol('d(theta1_dot)/dt'), (theta1_ddot_expr)))
print(sp.print_latex(theta1_ddot_expr))
display(Eq(sp.Symbol('d(theta2)/dt'), theta2_dot))
display(Eq(sp.Symbol('d(theta2_dot)/dt'), (theta2_ddot_expr)))
print(sp.print_latex(theta2_ddot_expr))
```

```python
import sympy as sp
from IPython.display import display
from sympy import Eq

# 定义符号变量
t = sp.Symbol('t')
# 直接代入数值，不再作为符号变量
m1_val, m2_val = 1, 1
L1_val, L2_val = 1, 1
# m1_val, m2_val = sp.symbols('m1 m2')
# L1_val, L2_val = sp.symbols('L1 L2')
g_val = 10
# c1, c2 = sp.symbols('c1 c2', positive=True) 
c1, c2 = 1e-1, 1e-1

# 定义角度和角速度作为时间 t 的函数
theta1 = sp.Function('theta1')(t)
theta2 = sp.Function('theta2')(t)
theta1_dot = sp.diff(theta1, t)
theta2_dot = sp.diff(theta2, t)
theta1_ddot = sp.diff(theta1_dot, t)
theta2_ddot = sp.diff(theta2_dot, t)

# 计算动能 T
T1 = sp.Rational(1, 2) * m1_val * (L1_val * theta1_dot)**2
T2 = sp.Rational(1, 2) * m2_val * (
    (L1_val * theta1_dot)**2 + (L2_val * theta2_dot)**2 + 2 * L1_val * L2_val * theta1_dot * theta2_dot * sp.cos(theta1 - theta2)
)
T = T1 + T2

# 计算势能 V
V = -((m1_val + m2_val) * g_val * L1_val * sp.cos(theta1) + m2_val * g_val * L2_val * sp.cos(theta2))

# 计算拉格朗日量 L
L = T - V

display(L)

# 计算瑞利耗散函数 F_D
FD = sp.Rational(1, 2) * c1 * theta1_dot**2 + sp.Rational(1, 2) * c2 * theta2_dot**2

# 使用欧拉-拉格朗日方程并考虑阻尼
eq1_lhs = sp.diff(sp.diff(L, theta1_dot), t) - sp.diff(L, theta1) + sp.diff(FD, theta1_dot)
eq2_lhs = sp.diff(sp.diff(L, theta2_dot), t) - sp.diff(L, theta2) + sp.diff(FD, theta2_dot)

display(eq1_lhs, eq2_lhs)

# 提取线性项，解出 theta1_ddot 和 theta2_ddot
A11 = sp.simplify(sp.diff(eq1_lhs, theta1_ddot))
A12 = sp.simplify(sp.diff(eq1_lhs, theta2_ddot))
A21 = sp.simplify(sp.diff(eq2_lhs, theta1_ddot))
A22 = sp.simplify(sp.diff(eq2_lhs, theta2_ddot))

B1 = sp.simplify(eq1_lhs - A11*theta1_ddot - A12*theta2_ddot)
B2 = sp.simplify(eq2_lhs - A21*theta1_ddot - A22*theta2_ddot)

M = sp.Matrix([[A11, A12], [A21, A22]])
RHS = sp.Matrix([B1, B2])

# 求解线性方程组 M * X = -RHS
solutions = M.solve(-RHS)

theta1_ddot_expr_damped = solutions[0]
theta2_ddot_expr_damped = solutions[1]

print("--- 有阻尼双摆一阶 ODE 系统 (dx/dt) ---")
display(Eq(sp.Symbol('d(theta1)/dt'), theta1_dot))
display(Eq(sp.Symbol('d(theta1_dot)/dt'), (theta1_ddot_expr_damped)))
print(sp.print_latex(theta1_ddot_expr_damped))
display(Eq(sp.Symbol('d(theta2)/dt'), theta2_dot))
display(Eq(sp.Symbol('d(theta2_dot)/dt'), (theta2_ddot_expr_damped)))
print(sp.print_latex(theta2_ddot_expr_damped))
```
